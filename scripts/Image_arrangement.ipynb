This script is to arrange the images into train, valid and test directories.

```python
from google.colab import drive
drive.mount('/content/drive')
path = '/content/drive/MyDrive/School/augmented_dataset/train'
```

```python
import os
import shutil
```

```python
uncracked_bridge_dir = os.path.join(path, 'uncracked_bridge')
cracked_bridge_dir = os.path.join(path, 'cracked_bridge')
destination = '/content/drive/MyDrive/School/augmented_dataset'
```

```python
# copying uncracked bridge images from uncracked_bridge_dir to destination
for filename in os.listdir(uncracked_bridge_dir):
    if filename.lower().endswith(('.jpg','.png','.jpeg')):
        src = os.path.join(uncracked_bridge_dir, filename)
        dst = os.path.join(destination, filename)
        shutil.copyfile(src, dst)
```

```python
# copying cracked bridge images from uncracked_bridge_dir to destination
for filename in os.listdir(cracked_bridge_dir):
    if filename.lower().endswith(('.jpg','.png','.jpeg')):
        src = os.path.join(cracked_bridge_dir, filename)
        dst = os.path.join(destination, filename)
        shutil.copyfile(src, dst)
```
