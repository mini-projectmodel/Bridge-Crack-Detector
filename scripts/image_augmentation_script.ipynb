```python
import matplotlib.pyplot as plt
import numpy as np
import random
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.image import ImageDataGenerator
%matplotlib inline
```

```python
import os
```

```python
# Define the base directory for augmented images in Google Drive
output_base_dir = '/content/drive/MyDrive/School/dataset'

# Define specific directories for uncracked and cracked augmented images
output_uncracked_dir = os.path.join(output_base_dir, 'uncracked_aug')
output_cracked_dir = os.path.join(output_base_dir, 'cracked_aug')

# Create the directories if they don't already exist
os.makedirs(output_uncracked_dir, exist_ok=True)
os.makedirs(output_cracked_dir, exist_ok=True)

print(f"Output directory for uncracked augmented images: {output_uncracked_dir}")
print(f"Output directory for cracked augmented images: {output_cracked_dir}")
```

```python
# This function will plot images in the form of a grid with 1 row and 10 columns where images are placed
def plotImages(images_arr):
    fig, axes = plt.subplots(1, 10, figsize=(20,20))
    axes = axes.flatten()
    for img, ax in zip(images_arr, axes):
        ax.imshow(img)
        ax.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
gen = ImageDataGenerator(rotation_range=10, width_shift_range=0.1, height_shift_range=0.1,
                         shear_range=0.15, zoom_range=0.1, channel_shift_range=10., horizontal_flip=True)
```

```python
dataset = '/content/drive/MyDrive/School/dataset'
```

```python
uncracked_image = dataset + '/uncracked'
cracked_image = dataset + '/cracked'
```

```python
# Copy original uncracked images to output_uncracked_dir with 'uncracked_' prefix
import shutil

print(f"Copying original uncracked images with prefix to: {output_uncracked_dir}")
for filename in os.listdir(uncracked_image):
    if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
        src_path = os.path.join(uncracked_image, filename)
        base_name, ext = os.path.splitext(filename)
        # Adding '_original' to differentiate these from actual augmented images
        dest_filename = f"uncracked_{base_name}_original{ext}"
        dest_path = os.path.join(output_uncracked_dir, dest_filename)
        shutil.copy(src_path, dest_path)
print("Finished copying original uncracked images.")
```

```python
# Copy original cracked images to output_cracked_dir with 'cracked_' prefix
print(f"Copying original cracked images with prefix to: {output_cracked_dir}")
for filename in os.listdir(cracked_image):
    if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
        src_path = os.path.join(cracked_image, filename)
        base_name, ext = os.path.splitext(filename)
        # Adding '_original' to differentiate these from actual augmented images
        dest_filename = f"cracked_{base_name}_original{ext}"
        dest_path = os.path.join(output_cracked_dir, dest_filename)
        shutil.copy(src_path, dest_path)
print("Finished copying original cracked images.")
```

```python
# Generate and save augmented images for 'uncracked' class
print(f"Generating augmented images for uncracked class and saving to: {output_uncracked_dir}")
for filename in os.listdir(uncracked_image):
    if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
        image_path = os.path.join(uncracked_image, filename)
        try:
            # Read image and expand dimensions for ImageDataGenerator
            image = tf.keras.preprocessing.image.load_img(image_path)
            image = tf.keras.preprocessing.image.img_to_array(image)
            image = np.expand_dims(image, 0)

            # Generate augmented images
            aug_iter = gen.flow(image, batch_size=1)
            num_aug_images = 20 # Number of augmented images per original image
            base_name, ext = os.path.splitext(filename)

            for i in range(num_aug_images):
                aug_image = next(aug_iter)[0].astype(np.uint8) # Get one augmented image
                # Modified: Prepend 'uncracked_' to the filename
                save_path = os.path.join(output_uncracked_dir, f"uncracked_{base_name}_aug_{i}{ext}")
                tf.keras.preprocessing.image.save_img(save_path, aug_image)
            # print(f"  Augmented {filename} and saved {num_aug_images} images.")
        except Exception as e:
            print(f"Error processing {filename}: {e}")
print("Finished generating uncracked augmented images.")
```

```python
# Generate and save augmented images for 'cracked' class
print(f"Generating augmented images for cracked class and saving to: {output_cracked_dir}")
for filename in os.listdir(cracked_image):
    if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
        image_path = os.path.join(cracked_image, filename)
        try:
            # Read image and expand dimensions for ImageDataGenerator
            image = tf.keras.preprocessing.image.load_img(image_path)
            image = tf.keras.preprocessing.image.img_to_array(image)
            image = np.expand_dims(image, 0)

            # Generate augmented images
            aug_iter = gen.flow(image, batch_size=1)
            num_aug_images = 20 # Number of augmented images per original image
            base_name, ext = os.path.splitext(filename)

            for i in range(num_aug_images):
                aug_image = next(aug_iter)[0].astype(np.uint8) # Get one augmented image
                # Modified: Prepend 'cracked_' to the filename
                save_path = os.path.join(output_cracked_dir, f"cracked_{base_name}_aug_{i}{ext}")
                tf.keras.preprocessing.image.save_img(save_path, aug_image)
            # print(f"  Augmented {filename} and saved {num_aug_images} images.")
        except Exception as e:
            print(f"Error processing {filename}: {e}")
print("Finished generating cracked augmented images.")
```
